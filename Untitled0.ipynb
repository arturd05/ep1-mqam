```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats

# Configuração visual similar aos plots do curso
sns.set_theme(style="darkgrid")

# 1. Carregamento dos dados
df_tread = pd.read_csv('treadwear.txt', sep='\t')
df_alli = pd.read_csv('alligator.txt', delim_whitespace=True)
df_patog = pd.read_csv('ConcPatog.txt', sep='\t')

# Função que gera tudo de uma vez
def analise_regressao(df, x_col, y_col, title, plot_hist=True, log_x=False, log_y=False, shift_zero=False):
    x_val = df[x_col].copy()
    y_val = df[y_col].copy()

    # Tratamento para zeros na base ConcPatog
    if shift_zero:
        x_val, y_val = x_val + 1, y_val + 1

    if log_x: x_val = np.log10(x_val)
    if log_y: y_val = np.log10(y_val)

    # Ajuste do Modelo
    X = sm.add_constant(x_val)
    model = sm.OLS(y_val, X).fit()
    print(f"\n==================== Resultados para: {title} ====================")
    print(model.summary().tables[0])
    print(model.summary().tables[1])

    # Painel de Gráficos
    fig = plt.figure(figsize=(14, 10))
    fig.suptitle(title, fontsize=16, fontweight='bold')

    # Gráfico 1: Dispersão com Linha de Regressão, IC e IP
    ax1 = fig.add_subplot(2, 2, 1)
    pred = model.get_prediction(X).summary_frame(alpha=0.05)

    sort_idx = np.argsort(x_val)
    x_sorted = x_val.iloc[sort_idx]

    ax1.scatter(x_val, y_val, alpha=0.7, color='steelblue', label='Dados Observados')
    ax1.plot(x_sorted, pred['mean'].iloc[sort_idx], color='crimson', linewidth=2, label='Reta de Regressão')
    ax1.fill_between(x_sorted, pred['mean_ci_lower'].iloc[sort_idx], pred['mean_ci_upper'].iloc[sort_idx], color='crimson', alpha=0.2, label='IC 95% (Média)')
    ax1.fill_between(x_sorted, pred['obs_ci_lower'].iloc[sort_idx], pred['obs_ci_upper'].iloc[sort_idx], color='gray', alpha=0.15, label='IP 95% (Predição)')
    ax1.set_xlabel(x_col + (' (log10)' if log_x else ''))
    ax1.set_ylabel(y_col + (' (log10)' if log_y else ''))
    ax1.set_title('Intervalos de Confiança e Predição')
    ax1.legend()

    # Gráfico 2: Resíduos vs Preditos
    ax2 = fig.add_subplot(2, 2, 2)
    ax2.scatter(model.fittedvalues, model.resid, alpha=0.7, color='purple')
    ax2.axhline(0, color='red', linestyle='--', linewidth=1.5)
    ax2.set_xlabel('Valores Preditos')
    ax2.set_ylabel('Resíduos')
    ax2.set_title('Resíduos vs Preditos')

    # Gráfico 3: QQ-Plot
    ax3 = fig.add_subplot(2, 2, 3)
    stats.probplot(model.resid, dist="norm", plot=ax3)
    ax3.set_title('QQ Plot dos Resíduos - Verificação de Normalidade')
    ax3.get_lines()[0].set_markerfacecolor('hotpink')
    ax3.get_lines()[0].set_markeredgecolor('hotpink')

    # Gráfico 4: Histograma (se ativado)
    if plot_hist:
        ax4 = fig.add_subplot(2, 2, 4)
        sns.histplot(model.resid, stat="density", kde=False, ax=ax4, color='skyblue')

        # Desenha a curva normal teórica
        mu, std = stats.norm.fit(model.resid)
        xmin, xmax = ax4.get_xlim()
        x = np.linspace(xmin, xmax, 100)
        p = stats.norm.pdf(x, mu, std)
        ax4.plot(x, p, 'k', linewidth=2, label='Distribuição Normal')
        ax4.axvline(0, color='red', linestyle='--')
        ax4.legend()
        ax4.set_xlabel('Resíduos')
        ax4.set_title('Histograma dos Resíduos')

    plt.tight_layout()
    plt.show()

# --- Etapas de Execução (Basta rodar o código) ---

# 1. Dataset: Treadwear (N=9, omitindo histograma)
analise_regressao(df_tread, 'mileage', 'groove', 'Treadwear - Original', plot_hist=False)
analise_regressao(df_tread, 'mileage', 'groove', 'Treadwear - Transf. (Somente Y)', plot_hist=False, log_y=True)

# 2. Dataset: Alligators
analise_regressao(df_alli, 'length', 'weight', 'Alligators - Original')
analise_regressao(df_alli, 'length', 'weight', 'Alligators - Transf. (Ambos X e Y)', log_x=True, log_y=True)

# 3. Dataset: ConcPatog
analise_regressao(df_patog, 'CPatog1', 'CPatog2', 'ConcPatog - Original')
analise_regressao(df_patog, 'CPatog1', 'CPatog2', 'ConcPatog - Transf. (Ambos X e Y)', log_x=True, log_y=True, shift_zero=True)
```
